```python
#prob2
import numpy as np
# 1. Create an array with values ranging from 10 to 49
arr1 = np.arange(10, 50)
print("1. Array from 10 to 49:")
print(arr1)
# 2. Create a 3X3 matrix with values ranging from 0 to 8
arr2 = np.arange(9).reshape(3, 3)
print("\n2. 3X3 Matrix:")
print(arr2)
# 3. Create a 3X3 identity matrix
arr3 = np.eye(3)
print("\n3. Identity Matrix:")
print(arr3)
# 4. Create a random array of size 30 and find the mean
arr4 = np.random.random(30)
print("\n4. Random Array:")
print(arr4)
print("Mean:", arr4.mean())
# 5. Create a 10X10 random array and find minimum and maximum
arr5 = np.random.random((10, 10))
print("\n5. 10X10 Random Array:")
print(arr5)
print("Minimum:", arr5.min())
print("Maximum:", arr5.max())
# 6. Create a zero array of size 10 and replace 5th element with 1
arr6 = np.zeros(10)
arr6[4] = 1
print("\n6. Zero Array with 5th element replaced:")
print(arr6)
# 7. Reverse an array
arr7 = np.array([1, 2, 0, 0, 4, 0])
print("\n7. Reversed Array:")
print(arr7[::-1])
# 8. Create a 2D array with 1 on border and 0 inside
arr8 = np.ones((5, 5))
arr8[1:-1, 1:-1] = 0
print("\n8. Border Array:")
print(arr8)
# 9. Create an 8X8 checkerboard pattern
arr9 = np.zeros((8, 8), dtype=int)
arr9[1::2, ::2] = 1
arr9[::2, 1::2] = 1
print("\n9. Checkerboard Pattern:")
print(arr9)
```

```python
#prob 4
import numpy as np
A = np.array([[3, 4], [7, 8]])
B = np.array([[5, 3], [2, 1]])
# 1. Prove A.A inverse = Identity Matrix
A_inverse = np.linalg.inv(A)
print("1. Inverse of A:")
print(A_inverse)
print("\nA multiplied by inverse of A:")
print(np.dot(A, A_inverse))
print("\nIdentity Matrix:")
print(np.eye(2))
print("A.A_inverse = I:", np.allclose(
    np.dot(A, A_inverse), np.eye(2)
))
# 2. Prove AB is not equal to BA
AB = np.dot(A, B)
BA = np.dot(B, A)
print("\n2. AB:")
print(AB)
print("\nBA:")
print(BA)
print("\nAB is not equal to BA:", not np.allclose(AB, BA))
# 3. Prove (AB)T = BT AT
left_side = np.transpose(np.dot(A, B))
right_side = np.dot(np.transpose(B), np.transpose(A))
print("\n3. (AB) Transpose:")
print(left_side)
print("\nB Transpose multiplied by A Transpose:")
print(right_side)
print("\nBoth sides are equal:",
      np.allclose(left_side, right_side))
# Solve the system of linear equations
# 2x - 3y + z = -1
# x - y + 2z = -3
# 3x + y - z = 9
coefficient = np.array([
    [2, -3, 1],
    [1, -1, 2],
    [3, 1, -1]
])
constants = np.array([-1, -3, 9])
# Solve using inverse method
print("\n4. Solution using inverse method:")
inverse = np.linalg.inv(coefficient)
solution = np.dot(inverse, constants)
print("x, y, z =", solution)
# Solve using np.linalg.inv()
print("\n5. Solution using np.linalg.inv:")
solution2 = np.linalg.inv(coefficient) @ constants
print("x, y, z =", solution2)
```

```python
#prob1
import numpy as np
import time
# Create two Python lists of size 1,000,000
list1 = list(range(1000000))
list2 = list(range(1000000))
# Python list addition
start = time.time()
result = [a + b for a, b in zip(list1, list2)]
end = time.time()
print("Python List Addition Time:", end - start, "seconds")
# Create NumPy arrays
arr1 = np.array(list1)
arr2 = np.array(list2)
# NumPy array addition
start = time.time()
result = arr1 + arr2
end = time.time()
print("NumPy Addition Time:", end - start, "seconds")
```
